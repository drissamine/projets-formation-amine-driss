## 2. Concepts fondamentaux de Keras

### 2.1 Architecture des réseaux de neurones

**Couches (Layers)** : Les composants de base des réseaux de neurones
- **Dense** : Couche entièrement connectée où chaque neurone reçoit des entrées de tous les neurones de la couche précédente
- **Activation** : Introduit la non-linéarité (ReLU, sigmoid, tanh, softmax)
- **Dropout** : Régularisation par désactivation aléatoire de neurones
- **BatchNormalization** : Normalise les activations pour stabiliser l'entraînement

**Fonctions d'activation principales**

| Activation | Formule | Utilisation |
|---|---|---|
| ReLU | max(0, x) | Couches cachées (défaut) |
| Sigmoid | 1/(1+e^(-x)) | Sortie binaire [0, 1] |
| Softmax | e^x / Σe^x | Sortie multi-classe |
| Tanh | (e^x - e^(-x))/(e^x + e^(-x)) | Alternatives aux couches cachées |
| Linear | x | Régression (pas d'activation) |

**Fonctions de perte (Loss functions)**

- **Classification binaire** : `binary_crossentropy`
- **Classification multi-classe** : `categorical_crossentropy` ou `sparse_categorical_crossentropy`
- **Régression** : `mean_squared_error`, `mean_absolute_error`

**Optimiseurs**

- **SGD** : Stochastic Gradient Descent basique
- **Adam** : Adaptation du taux d'apprentissage adaptatif (recommandé, défaut)
- **RMSprop** : Root Mean Square Propagation
- **Adagrad** : Pour données creuses


## 3. Modèle Séquentiel - Classification (Iris)

### 3.1 Théorie du modèle séquentiel

Le **modèle séquentiel** est le type de modèle le plus simple dans Keras. Les couches sont empilées linéairement une après l'autre.

```python
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(4,)),
    layers.Dense(32, activation='relu'),
    layers.Dense(3, activation='softmax')
])
```

**Avantages** :
- Simple et intuitif
- Facile à construire et visualiser
- Approprié pour 90% des cas

**Limitations** :
- Une seule entrée, une seule sortie
- Pas de sauts de connexions (skip connections)
- Pas de structure branchée

## 4. Callbacks : Contrôle de l'entraînement

### 4.1 Théorie des callbacks

Les **callbacks** sont des fonctions appelées à des moments spécifiques pendant l'entraînement. Ils permettent de :

1. **Early Stopping** : Arrêter l'entraînement si la validation ne s'améliore plus
2. **Model Checkpoint** : Sauvegarder le meilleur modèle
3. **Learning Rate Reduction** : Réduire le taux d'apprentissage dynamiquement
4. **Custom Callbacks** : Implémenter des comportements personnalisés

### 4.2 Implémentation des callbacks

````python
# Définition des callbacks

# Early Stopping: arrête l'entraînement si val_loss ne s'améliore pas pendant 10 epochs
early_stopping = callbacks.EarlyStopping(
    monitor='val_loss',           # Métrique à surveiller
    patience=10,                  # Nombre d'epochs sans amélioration
    restore_best_weights=True,    # Restaurer les poids du meilleur modèle
    verbose=1
)

# Model Checkpoint: sauvegarde le meilleur modèle
model_checkpoint = callbacks.ModelCheckpoint(
    'best_model.h5',              # Chemin de sauvegarde
    monitor='val_accuracy',       # Métrique à surveiller
    save_best_only=True,          # Sauvegarder uniquement le meilleur
    verbose=0
)

# Réduction du taux d'apprentissage
reduce_lr = callbacks.ReduceLROnPlateau(
    monitor='val_loss',           # Métrique à surveiller
    factor=0.5,                   # Multiplier le taux par 0.5
    patience=5,                   # Après 5 epochs sans amélioration
    min_lr=1e-6,                  # Taux minimum
    verbose=1
)

print("Callbacks définis:")
print("- Early Stopping")
print("- Model Checkpoint")
print("- Learning Rate Reduction")
````

## 5. Modèle Fonctionnel - Classification avancée

### 5.1 Théorie du modèle fonctionnel

Le **modèle fonctionnel** offre plus de flexibilité que le modèle séquentiel :

**Avantages** :
- Plusieurs entrées/sorties
- Connexions non-linéaires (skip connections)
- Architectures complexes (Inception, ResNet, etc.)

**Syntaxe** :
```python
inputs = keras.Input(shape=(4,))
x = layers.Dense(64, activation='relu')(inputs)
x = layers.Dense(32, activation='relu')(x)
outputs = layers.Dense(3, activation='softmax')(x)
model = keras.Model(inputs=inputs, outputs=outputs)
```
