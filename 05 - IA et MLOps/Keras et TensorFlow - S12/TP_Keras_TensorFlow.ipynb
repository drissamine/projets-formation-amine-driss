````python
# Fonction utilitaire pour afficher les courbes d'entraînement
def plot_history(history, title=""):
    """Affiche les courbes de loss et de métrique pour le train et la validation."""
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Loss
    axes[0].plot(history.history['loss'], label='Train Loss', linewidth=2)
    axes[0].plot(history.history['val_loss'], label='Val Loss', linewidth=2)
    axes[0].set_title(f'Loss — {title}')
    axes[0].set_xlabel('Époque')
    axes[0].set_ylabel('Loss')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)
    
    # Métrique (accuracy ou mae)
    metric_keys = [k for k in history.history.keys() if k not in ['loss', 'val_loss', 'lr']]
    train_metric = [k for k in metric_keys if not k.startswith('val_')][0]
    val_metric = f'val_{train_metric}'
    
    axes[1].plot(history.history[train_metric], label=f'Train {train_metric}', linewidth=2)
    axes[1].plot(history.history[val_metric], label=f'Val {train_metric}', linewidth=2)
    axes[1].set_title(f'{train_metric} — {title}')
    axes[1].set_xlabel('Époque')
    axes[1].set_ylabel(train_metric)
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
````

#  PARTIE 2 — Les Callbacks
---

Les **callbacks** sont des fonctions appelées automatiquement à différents moments de l'entraînement (début/fin d'époque, de batch, etc.).

### Callbacks les plus courants :

| Callback | Description |
|---|---|
| `EarlyStopping` | Arrête l'entraînement si la métrique surveillée ne s'améliore plus |
| `ModelCheckpoint` | Sauvegarde le meilleur modèle pendant l'entraînement |
| `ReduceLROnPlateau` | Réduit le learning rate quand la loss stagne |
| `TensorBoard` | Permet la visualisation avec TensorBoard |
| `CSVLogger` | Enregistre les logs dans un fichier CSV |
| **Custom Callback** | Callback personnalisé via héritage de `keras.callbacks.Callback` |

```python
class TrainingMonitor(keras.callbacks.Callback):
    def on_train_begin(self, logs=None):
        self.start_time = time.time()
        self.prev_val_loss = None
        print("🚀 Début de l'entraînement...")
        print("=" * 60)
    
    def on_epoch_end(self, epoch, logs=None):
        val_loss = logs.get('val_loss', 0)
        
        # Affichage toutes les 10 époques
        if (epoch + 1) % 10 == 0:
            print(f"  📊 Époque {epoch+1:3d} | "
                  f"Loss: {logs['loss']:.4f} | "
                  f"Val Loss: {val_loss:.4f} | "
                  f"Val Acc: {logs.get('val_accuracy', 0):.4f}")
        
        # Warning si val_loss augmente fortement
        if self.prev_val_loss is not None and val_loss > self.prev_val_loss * 1.1:
            print(f"  ⚠️  Époque {epoch+1}: val_loss a augmenté de "
                  f"{((val_loss/self.prev_val_loss)-1)*100:.1f}%!")
        
        self.prev_val_loss = val_loss
    
    def on_train_end(self, logs=None):
        total_time = time.time() - self.start_time
        print("=" * 60)
        print(f"🏁 Entraînement terminé en {total_time:.2f} secondes")
```

L'API **fonctionnelle** de Keras permet de créer des architectures plus complexes :
- **Entrées multiples** (ex: texte + image)
- **Sorties multiples** (ex: classification + régression)
- **Couches partagées** (ex: réseaux siamois)
- **Connexions résiduelles** (skip connections)
- **Graphes non-linéaires**

### Syntaxe :
```python
# Séquentiel           vs          Fonctionnel
model = Sequential([               inputs = layers.Input(shape=(n,))
    layers.Dense(64),               x = layers.Dense(64)(inputs)
    layers.Dense(32),               x = layers.Dense(32)(x)
    layers.Dense(1)                 outputs = layers.Dense(1)(x)
])                                  model = models.Model(inputs, outputs)
```

### Chargement des données

````python
# Fashion MNIST est directement disponible dans Keras (originalement de Kaggle)
(X_train_f, y_train_f), (X_test_f, y_test_f) = keras.datasets.fashion_mnist.load_data()

class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

print(f"Train : {X_train_f.shape}, Test : {X_test_f.shape}")
print(f"Nombre de classes : {len(class_names)}")

# Visualiser quelques exemples
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train_f[i], cmap='gray')
    ax.set_title(class_names[y_train_f[i]], fontsize=10)
    ax.axis('off')
plt.suptitle('Exemples Fashion MNIST', fontsize=14)
plt.tight_layout()
plt.show()
````

#  PARTIE 5 — Résumé et Tableau Comparatif

## Tableau récapitulatif

| Concept | Description | Quand l'utiliser |
|---------|-------------|------------------|
| **Sequential API** | Couches empilées linéairement | Prototypage rapide, architectures simples |
| **Functional API** | Graphe de couches flexible | Multi-input/output, skip connections, architectures complexes |
| **Dropout** | Désactive aléatoirement des neurones | Réduire l'overfitting (taux typique : 0.2-0.5) |
| **EarlyStopping** | Arrête l'entraînement quand la val_loss stagne | Toujours recommandé |
| **ModelCheckpoint** | Sauvegarde le meilleur modèle | Entraînements longs, production |
| **ReduceLROnPlateau** | Réduit le learning rate dynamiquement | Quand la convergence ralentit |
| **Custom Callback** | Logique personnalisée pendant l'entraînement | Logging, alertes, intégration MLOps |
| **BatchNormalization** | Normalise les activations entre couches | Stabiliser et accélérer l'entraînement |
| **Skip Connection** | Connexion résiduelle entre couches | Réseaux profonds, éviter vanishing gradient |
